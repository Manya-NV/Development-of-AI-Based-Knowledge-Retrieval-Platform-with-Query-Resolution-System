## Question 1: Class and Object

### Prompt
Explain the difference between a class and an object in Python. Write a Python program that defines a Student class with name and age attributes and a method to display student details. Explain the code and its output.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)


student1 = Student("Manya", 21)
student1.display()
```

### Explanation

A class is a blueprint used to create objects. An object is an instance of a class.

- `Student` is the class.
- `student1` is an object of the Student class.
- `__init__()` initializes the object's attributes when it is created.
- `self` refers to the current object.
- `display()` is an instance method that prints the student's details.

### Time Complexity
O(1) for creating and displaying this fixed number of attributes.

### Space Complexity
O(1) for this single object with a fixed number of attributes.

## Question 2: Constructor in Python

### Prompt
Explain the purpose of a constructor in Python. Create an Employee class whose constructor initializes the employee's name and salary. Display the employee details and explain the output.

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display(self):
        print("Employee:", self.name)
        print("Salary:", self.salary)


emp = Employee("Rahul", 30000)
emp.display()
```

### Output
Employee: Rahul
Salary: 30000

### Explanation
The `__init__()` method is automatically called when an object is created. It initializes the object's attributes using the supplied values.

### Time Complexity
O(1) for this example.

### Space Complexity
O(1) for one employee object.

## Question 3: Instance Variables and Class Variables

### Prompt
Explain the difference between instance variables and class variables in Python. Create a Student class with a class variable representing the college name and instance variables representing each student's name. Create two objects and display their details.

```python
class Student:
    college = "Malnad College of Engineering"

    def __init__(self, name):
        self.name = name

    def display(self):
        print("Name:", self.name)
        print("College:", Student.college)


student1 = Student("Anu")
student2 = Student("Ravi")

student1.display()
student2.display()
```

### Output
Name: Anu
College: Malnad College of Engineering
Name: Ravi
College: Malnad College of Engineering

### Explanation
An instance variable belongs to an individual object. A class variable is shared through the class and is commonly used for data that applies to all instances.

Here, each student has a separate name, but both use the same college name.

### Time Complexity
O(1) per display call for this fixed example.

### Space Complexity
O(1) per object, excluding implementation overhead.

## Question 4: Encapsulation

### Prompt
Explain encapsulation in Python. Create a BankAccount class with a private balance attribute, and provide methods to deposit money and display the balance. Demonstrate how these methods work.

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def display_balance(self):
        print("Balance:", self.__balance)


account = BankAccount(1000)
account.deposit(500)
account.display_balance()
```

### Output
Balance: 1500

### Explanation
Encapsulation combines data and the methods that operate on it within a class. The `__balance` attribute uses Python's name-mangling convention to discourage direct access. The deposit method changes the balance only when the amount is positive.

Python's double underscore does not provide absolute security.

### Time Complexity
O(1) for the deposit and display operations.

### Space Complexity
O(1) for the account object.

## Question 5: Inheritance

### Prompt
Explain inheritance in Python. Create a parent class Animal with a method named eat(), and a child class Dog that inherits from Animal. Add a bark() method to Dog and demonstrate both methods.

```python
class Animal:
    def eat(self):
        print("Animal is eating")


class Dog(Animal):
    def bark(self):
        print("Dog is barking")


dog = Dog()
dog.eat()
dog.bark()
```

### Output
Animal is eating
Dog is barking

### Explanation
Inheritance allows a child class to reuse methods and attributes from a parent class. `Dog` inherits the `eat()` method from `Animal` and defines its own `bark()` method.

### Time Complexity
O(1) for these method calls.

### Space Complexity
O(1) for this single object, excluding implementation overhead.

## Question 6: Multilevel Inheritance

### Prompt
Explain multilevel inheritance in Python. Create three classes: Animal, Dog, and Puppy, where Dog inherits from Animal and Puppy inherits from Dog. Demonstrate how a Puppy object can use methods from all three classes.

```python
class Animal:
    def eat(self):
        print("Animal eats")


class Dog(Animal):
    def bark(self):
        print("Dog barks")


class Puppy(Dog):
    def weep(self):
        print("Puppy weeps")


puppy = Puppy()
puppy.eat()
puppy.bark()
puppy.weep()
```

### Output
Animal eats
Dog barks
Puppy weeps

### Explanation
Multilevel inheritance occurs when one class inherits from a class that itself inherits from another class. `Puppy` inherits `bark()` from `Dog` and `eat()` from `Animal`, while also defining `weep()`.

### Time Complexity
O(1) for these method calls.

### Space Complexity
O(1) for this single object, excluding implementation overhead.

## Question 7: Polymorphism

### Prompt
Explain polymorphism in Python. Create two classes, Cat and Dog, that each implement a sound() method. Write a program that calls sound() on objects of both classes using the same method name.

```python
class Cat:
    def sound(self):
        print("Cat says Meow")


class Dog:
    def sound(self):
        print("Dog says Woof")


animals = [Cat(), Dog()]

for animal in animals:
    animal.sound()
```

### Output
Cat says Meow
Dog says Woof

### Explanation
Polymorphism means that different objects can respond to the same method call in their own ways. Both classes implement `sound()`, but each produces different behavior.

### Time Complexity
O(1) per call in this example.

### Space Complexity
O(1) for the fixed number of objects.

## Question 8: Method Overriding

### Prompt
Explain method overriding in Python. Create a parent class Vehicle with a start() method and a child class Car that overrides start(). Demonstrate how the child class implementation is called.

```python
class Vehicle:
    def start(self):
        print("Vehicle is starting")


class Car(Vehicle):
    def start(self):
        print("Car is starting")


vehicle = Vehicle()
car = Car()

vehicle.start()
car.start()
```

### Output
Vehicle is starting
Car is starting

### Explanation
Method overriding occurs when a child class provides its own implementation of a method already defined in the parent class. When `start()` is called on a `Car` object, Python uses the child's implementation.

### Time Complexity
O(1) per call in this example.

### Space Complexity
O(1) for these fixed objects.

## Question 9: Abstraction

### Prompt
Explain abstraction in Python. Use the abc module to create an abstract class Shape with an abstract method area(). Implement the method in a Rectangle class and display the rectangle's area.

```python
from abc import ABC, abstractmethod


class Shape(ABC):
    @abstractmethod
    def area(self):
        pass


class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


rectangle = Rectangle(5, 4)
print("Area:", rectangle.area())
```

### Output
Area: 20

### Explanation
Abstraction exposes essential behavior while hiding implementation details. `Shape` defines an abstract `area()` method. A concrete subclass such as `Rectangle` must implement that method before its objects can be created.

### Time Complexity
O(1) for calculating the area of a rectangle.

### Space Complexity
O(1) for the rectangle object.
